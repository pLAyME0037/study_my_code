```python
def firstN(n):
    i = 0
    while (i < n):
        yield i
        i += 1

print(list(firstN(10)))
```

```python
def add(a, b):
    return a + b
    
def sub(a, b):
    return a - b
    
print(add(4, 5))
print(add(6, 5))
print(add(6, 12))

print(sub(5, 5))
print(sub(6, 5))
print(sub(6, 12))
```
